THE PERCEPTRON

it is one of the simplest ANN arch invented in 1957, it is based on a little different neuron see fig 10-4 it is called threshold logic unit TLU or linear threshold unit LTU

the input and output here are numbers instead of on and off, and each number have some weights, the tlu first computes a linear function of its inputs

z = w1x1 + w2x2...... + wnxn + b = W^tX + b, then it applies a step function to the result
h(w)(x) = step(z) 

its like logistic regression but uses a step function additionaly

the most common step function used in perceptrons are heavside step function, and sometimes sign function see copy 1

a single tlu can perform binary classification task, it computes a linear function of its input and if its greater than a threshold it outputs a positive class, otherwise a negative class

a perceptron contains one or more TLU and each input is connected to each tlu see fig 10-5 such a connection layer is called fully connected layer or dense layer, the inputs is called input layer
the layer of tlus which gives the outputs it is called output layer see fig 10-5
which shows a two input and 3 output neurons perceptron arch
this perceptron can do a three class classification or multiclass classification

W is the weight matrix, row represents one input and column per neuron

ϕ is activation function, when the neurons are tlu it is a step function 

training :

a perceptron is training is largely inspired by HEbbs rule, when a biological neuron triggers another neuron the bond between them grows stronger, the cell that fire together wire together, this means the connection weights between them tends to increase when they are fired simultaneously, this rule later becomes hebbs rule  or hebbian learning, 

the perceptron is trained using a derivative of this rule which takes the error made during the prediction by the network into account and tries to always minimize the error by reinforcing the connections that help reduce the errror
the perceptron is given one training instance at a time and for each instance it makes its prediction and sees where that output neurons are producing wrong prediction, then it reinforces the connection weights from the inputs that may have contributed to the correct prediction,

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.linear_model import Perceptron #perceptron class

iris = load_iris(as_frame=True) #as_frame=True returns a pandas dataframe instead of a numpy array

X = iris.data[["petal length (cm)", "petal width (cm)"]].values
y = (iris.target == 0) # Iris setosa

per_clf = Perceptron(random_state=42)
per_clf.fit(X, y)
X_new = [[2, 0.5], [3, 1]]
y_pred = per_clf.predict(X_new) # predicts True and False for these 2 flowers
```

THE MULTILAYER PERCEPTRON AND BACKPROPOGATION

IT contains one input layer and one or two layers of TLU called hidden layers, and one final output layer, 
the layers close to input layer is called lower layers and one close to output layer is called upper layers see fig 10-7

the signal flows into one direction only so this arch is an example of feedforward neural network

when an ANN have multiple deep stack of hidden layers it is called deep neural network, researchers struggled a lot for finding how to train a MLP some said using gradient descent but this will require calculating the gradients of the model's error with regard to model params , but it was a difficult task to do with the computing they have at that time

then in 1970 a researcher formualated a technique to compute all the gradients automatically it is called reverse mode automatic differentiation, it can compute the gradients of the model with respect to every params using only two passes one forward and one backword
it can find out how the weights and bias of the connections can be modified so that network's error can reduce
these gradients can be used to perform a gradient descent step now and if we repeat it automatically taking a step, the model's error will reduce eventually until it reaches a minimum,
this combination of reverse mode diff and gradient descent is called backpropogation

there are various auto diff techniques each having its pros and cons, reverse auto diff is particularly used when there are many diff variables like connection weights and biases but only a few outputs 

backpropogation can be applied to all type of computational graphs, it is not meant for training neural netqorks only, it was a more generalised technique, the technique was not in mainstream for training the nn, but a new research showed how backpropogation can show some useful internal representations to the neural network, then it become quickly popularized

working:

it takes one mini batch at a time, for eg 16 instances at once, and goes to full train set multiple times, each pass is called an epoch

each mini batch enters into the network from input layer and the algorithm then computes the output of all neurons present in hidden layer for each instance, suppose we have 3 insances and 4 neuron so each instance will go through the 4 neurons and an output is produced.
the result is passed to next layer its output is calculated and then passed forward, this happens until final output is produced from the output layer
this is just like forward pass but here we save the results as we will need it for backpropogation

then the models error is calculated using the actual output and the expected output using some loss function

then it computes how much each output bias and connection contributed to the error and this is done by using chain rule of calculus

the algorithm then measures how much of these error contributions came from below layer again using the chain rule , this reverse pass calculates the gradients for all connnection weights and bias by propogating the error gradient backwards

finally another gradient descent step is performed to tweak all the connections weights in the network using the error gradients it calculated

it is important to randomly initialize the hidden layers connection weights and biases, if we initialize all of them as 0 then all the neurons in a given layer will be seen only as one neuron everyone will be identical, there will be no chance for the algorithm to adapt and form a diverse team of neurons

so in short backpropogation makes predictions for a mini batch in forward pass and then goes back to see where the error develops or to measure the error contribution of params in a reverse pass and at last tweaks the connection weights and biases to reduce the error (gradient descent step)

to make the back prop work properly researchers made some changes in MLP arch, 
they replaced the step function with the logistic function 
σ(z) = 1 / (1 + exp(–z)) 
also known as sigmoid function, one of the main reasons for doing this was the step function does not have any gradient to work with only flat segments and radient descent cannot move on flat sirfaces, 
sigmoid func provides non zero derivatives everywhere, allowing GD to make some progress at each step, 
the algo can work well with many other activation functions like 

hyperbolic tangent func tanh = 2σ(2z) – 1
just like sigmoid this is also S shaped, continuous and differentiable but its output ranges from -1 to 1 unlike 0 to 1 in sigmoid, this way each of the utput produces are centered more or less around 0 only at beginning of training which helps with speed up convergence

Rectified linear unit function ReLU(z) = max(0, z)
it is continuous but not differentiable at z=0, as the slope changes abruptly causing GD to bounce around and its derivative is 0 for z<0, but in practice it works good in fact it has the advantage of being fast to compute so it is now default, and the fact that it does not have a maximum output value helps reduce some errors during gradient descent as well
