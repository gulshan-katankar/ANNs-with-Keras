the ANNs were first introduced in 1943 by Warren McCulloch and the mathematician Walter Pitts, they presented a simplified computational model of how biological may works together to perform complex operations in animal brains using propositional logic, this was the first ANN arch

in 1980s new arch and better training methods were developed which sparked the interest in study of neural networks, but it was slow and by the 1990s SVMs were developed which are far more powerful than the NNs of previous times, they provided better results and stronger theoretical foundations than ANNs, so the study of ANNs was stopped again

biological neurons

see fig 10-1 it shows a cell found in animal brains, it contains cell body and the nucleus, many branching components called dendrites, plus one long extension called axon which is few times longer than the cell body and sometimes 100s or 1000 times long.

the axon splits into many branches called telodendria at its end point, and at the tips of these branches small structures called synaptic terminals or synapese is present which are furhter connected to the cell body or the dendrites of the next neuron

biological neurons produces a small electric impulses called action potentials or just signals which travel through the cell body and axon and reaches the synapse where it gets accumulated as a form of chemical signals released by synapse called neurotransmitters, and when there are enough of these collected (which takes few milliseconds), it fires its own electrical impulses, but it also depends on the type of neurotransmitters as some of them inhibit the neuron from firing 

the biological neuron behave pretty simple, but each neuron is connected to 1000s other its a complex network of billions of neurons and can perform complex computations
